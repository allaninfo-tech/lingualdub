## Transfer Opportunity Analysis

Runyankole and Luganda share Bantu morphology and geographic contact history.
Key conclusions from the audit:
1. **Acoustic Transfer**: Luganda acoustic checkpoints provide an effective warm-start for Runyankole ASR adaptation.
2. **Lexical Overlap**: High cognate overlap (~70-80%) allows shared tokenization strategies.
3. **Pretrained Checkpoints**: Meta MMS-TTS (`facebook/mms-tts-nyn`) and NLLB-200 (`nyn_Latn`) confirm zero-shot baseline viability without training from scratch.
