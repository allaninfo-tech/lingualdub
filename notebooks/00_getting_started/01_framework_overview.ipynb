# LingualDub — Framework Overview

This notebook walks through the core objects of the LingualDub framework:
Language, Resource, Component, Pipeline, and Result.

**Goal:** Understand how the pieces fit together before building anything.

## Setup

```python
# Install from local source if not already installed
# !pip install -e ../..

import lingualdub
print(f"LingualDub version: {lingualdub.__version__}")
```

## 1. Language

A `Language` describes what a language is and what resources and tasks it supports.
The resource profile is the first thing to look at — it tells you what you're working with.

```python
from lingualdub.core import Language

# Load the built-in Luganda profile
from lingualdub.languages.luganda import LUGANDA
print(LUGANDA)
print(f"Resource profile: {LUGANDA.resource_profile}")
print(f"Supported tasks:  {LUGANDA.supported_tasks}")
print(f"Related languages: {LUGANDA.related_languages}")
```

## 2. Resource

A `Resource` is any data asset — audio, text, checkpoints, eval sets.
Provenance is mandatory. It's what makes runs reproducible and comparable.

```python
from lingualdub.core import Resource
from lingualdub.core.resource import ResourceKind

# Example: define a speech resource
resource = Resource(
    id="lug_speech_example",
    kind=ResourceKind.SPEECH,
    language="lug",
    version="1.0.0",
    provenance={
        "source": "...",
        "license": "...",
        "collected_by": "..."
    }
)
print(resource)
```

## 3. Registry

The Registry is how everything gets discovered and resolved without hard-coding.

```python
from lingualdub.registry import Registry

registry = Registry()

# Register the language
registry.register("language", "lug", LUGANDA, version="1.0.0")
registry.register("resource", "lug_speech_example", resource, version="1.0.0")

print("Registered languages:", registry.list("language"))
print("Registered resources:", registry.list("resource"))
```

## 4. Result and Segment

Every component outputs a `Result`. Results carry `Segment` objects — the atomic
unit of speech/text data — plus status, warnings, and provenance.

```python
from lingualdub.core import Segment, Result
from lingualdub.core.result import ResultStatus

seg = Segment(start=0.5, end=3.2, text="Oli otya", language="lug", speaker="spk_01", confidence=0.91)
print(f"Segment: {seg}")
print(f"Duration: {seg.duration:.2f}s")

result = Result(segments=[seg], source_language="lug")
print(f"\nResult: {result}")
print(f"Usable: {result.is_usable}")
```
